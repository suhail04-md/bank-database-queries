1) Print customer Id, customer name and average account_balance maintained by each customer for all  of his/her accounts in the bank.(8 Rows)

```python
query="""SELECT
    bc.customer_id,
    bc.customer_name,
    AVG(bad.Balance_amount) AS average_balance
FROM BANK_CUSTOMER bc
JOIN Bank_Account_Details bad ON bc.customer_id = bad.Customer_id
GROUP BY bc.customer_id, bc.customer_name
LIMIT 8;"""

result = pd.read_sql_query(query, conn)
print(tabulate(result, headers='keys', tablefmt='psql'))
```

2) Print customer_id ,account_number and balance_amount , condition that if balance_amount is nil then assign transaction_amount  for account_type = "Credit Card"(4 Rows)

```python
query = """SELECT bad.Customer_id,
    bad.Account_Number,
    CASE
        WHEN bad.Balance_amount = 0 AND bad.Account_type IN ('Credit Card', 'Add-on Credit Card')
        THEN COALESCE((SELECT SUM(bat.Transaction_amount)
                       FROM BANK_ACCOUNT_TRANSACTION bat
                       WHERE bat.Account_Number = bad.Account_Number), 0)
        ELSE bad.Balance_amount
    END AS balance_amount
FROM Bank_Account_Details bad
WHERE bad.Account_type IN ('Credit Card', 'Add-on Credit Card')
LIMIT 4;"""

result = pd.read_sql_query(query, conn)
print(tabulate(result, headers='keys', tablefmt='psql'))
```

3) Print account_number and balance_amount , transaction_amount,Transaction_Date from Bank_Account_Details andbank_account_transaction for all the transactions occurred during march,2020 and april, 2020(12 Rows)

```python
query = """SELECT
    bad.Account_Number,
    bad.Balance_amount,
    bat.Transaction_amount,
    bat.Transaction_Date
FROM Bank_Account_Details bad
JOIN BANK_ACCOUNT_TRANSACTION bat ON bad.Account_Number = bat.Account_Number
WHERE strftime('%m', bat.Transaction_Date) IN ('03', '04')
  AND strftime('%Y', bat.Transaction_Date) = '2020'
  LIMIT 12;"""
result = pd.read_sql_query(query, conn)
print(tabulate(result, headers='keys', tablefmt='psql'))
```

4) Print all of the customer id, account number,  balance_amount, transaction_amount , Transaction_Datefrom bank_customer, Bank_Account_Details and bank_account_transaction tables where excluding all of their transactions in march, 2020  month (22 Rows)

```python
query = """SELECT
    bad.Customer_id,
    bad.Account_Number,
    bad.Balance_amount,
    bat.Transaction_amount,
    bat.Transaction_Date
FROM Bank_Account_Details bad
LEFT JOIN BANK_ACCOUNT_TRANSACTION bat ON bad.Account_Number = bat.Account_Number
    AND NOT (strftime('%m', bat.Transaction_Date) = '03' AND strftime('%Y', bat.Transaction_Date) = '2020')
    LIMIT 22;
"""
result = pd.read_sql_query(query, conn)
print(tabulate(result, headers='keys', tablefmt='psql'))
```

7) Print all Customer_id, Account_Number, Account_type, and display deducted balance_amount by  subtracting only negative transaction_amounts for Relationship_type ="P" ( P - means  Primary , S - means Secondary ) .(27 Rows)

```python
query = """SELECT
    bad.Customer_id,
    bad.Account_Number,
    bad.Account_type,
    bad.Balance_amount AS deducted_balance
FROM Bank_Account_Details bad
WHERE bad.Relationship_type = 'P'

UNION ALL

SELECT
    bad.Customer_id,
    bad.Account_Number,
    bad.Account_type,
    bad.Balance_amount - ABS(bat.Transaction_amount) AS deducted_balance
FROM Bank_Account_Details bad
JOIN BANK_ACCOUNT_TRANSACTION bat ON bad.Account_Number = bat.Account_Number
WHERE bad.Relationship_type = 'P'
  AND bat.Transaction_amount < 0
  LIMIT 27;"""
result = pd.read_sql_query(query, conn)
print(tabulate(result, headers='keys', tablefmt='psql'))
```

```python
query = """SELECT
    bard_savings.Account_Number AS Savings_Account,
    bard_cc.Account_Number AS Credit_Card_Account,
    SUM(bat.Transaction_amount) AS Total_Credit_Card_Transactions
FROM Bank_Account_Relationship_Details bard_savings
JOIN Bank_Account_Relationship_Details bard_cc
    ON bard_savings.Account_Number = bard_cc.Linking_Account_Number
JOIN BANK_ACCOUNT_TRANSACTION bat
    ON bard_cc.Account_Number = bat.Account_Number
WHERE bard_savings.Account_type = 'SAVINGS'
  AND bard_cc.Account_type = 'Credit Card'
GROUP BY bard_savings.Account_Number, bard_cc.Account_Number;"""
result = pd.read_sql_query(query, conn)
print(tabulate(result, headers='keys', tablefmt='psql'))
```
